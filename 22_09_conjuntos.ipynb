```python
# 1. Crie um conjunto chamado numeros contendo os números de 1 a 5. Mostre o conjunto.

números = {1,2,3,4,5}
print(números)
```

```python
# 2. Crie numeros = {1, 2, 2, 3, 3, 3, 4, 5}. Mostre o conjunto e observe o que acontece com os números repetidos.

números = {1, 2, 2, 3, 3, 3, 4, 5}
print(números)
```

```python
# 3. Crie frutas = {"maçã", "banana", "laranja"}. Adicione "morango" usando add().

frutas = {"maçã", "banana", "laranja"}
frutas.add("morango")
print(frutas)
```

```python
# 4. Crie cores = {"vermelho", "azul", "verde", "amarelo"}. Remova "verde" usando remove().

cores = {"vermelho", "azul", "verde", "amarelo"}
cores.remove("verde")
print(cores)
```

```python
# 5. Crie a lista: nomes = ["Yris", "Luna", "Luna", "Maitê", "Yris", "Ravi", “Gael",”Theo”, “Theo”]
#Transforme a lista em um conjunto para eliminar os nomes repetidos.

nomes = ["Yris", "Luna", "Luna", "Maitê", "Yris", "Ravi", "Gael", "Theo", "Theo"]
nomes_sem_repetir = set(nomes)
print("Como conjunto:", nomes_sem_repetir)

lista_limpa = list(nomes_sem_repetir)
print("Lista sem repetidos:", lista_limpa)
```
