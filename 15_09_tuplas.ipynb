```python
# 1. Crie uma tupla chamada meses contendo Janeiro, Fevereiro, Março e Abril. Mostre a tupla.

meses = ("Janeiro", "Fevereiro"," Março", "Abril")
print(meses)
```

```python
# 2. Crie numeros = (10, 20, 30, 40, 50). Mostre o primeiro, o terceiro e o último número.

números = (10, 20, 30, 40, 50)
print(números[0])
print(números[2])
print(números[4])
```

```python
# 3. Crie uma tupla aluno contendo o nome, a idade e o curso de um aluno. Mostre cada informação separadamente.

aluno = ("Melissa", 15, "Informática")
print(aluno)
```

```python
# 4. Crie cores = ("vermelho", "azul", "verde"). Tente alterar "azul" para "amarelo". Observe o que acontece.

cores = ("vermelho", "azul", "verde")
cores[1] = "amarelo"
print(cores)
```

```python
# 5. Crie numeros = (2, 4, 2, 6, 2, 8, 10). Use count() para descobrir quantas vezes o número 2 aparece.

números = (2, 4, 2, 6, 2, 8, 10)
quantidade = números.count(2)
print(quantidade)
```
