```python
#Exercício 1 — Contagem
#Faça um programa que peça um número ao usuário e mostre todos os números de 1 até esse número usando while.

a = int(input("digite um número: "))

contador = 1

while contador <= a:
    print(contador)
    contador += 1
```

```python
#Exercício 2 — Contagem regressiva
#Peça um número ao usuário e faça uma contagem até 0, mostrando cada número na tela.

n = int(input("digite um número: "))

while n >= 0:
  print(n)
  n -= 1
```

```python
#Exercício 3 — Números pares
#Faça um programa que mostre todos os números pares de 0 até 20 usando while.

n = 0

while n <= 20:
  print(n)
  n += 2
```

```python
#Exercício 4 — Soma dos números
#Peça um número ao usuário e calcule a soma de todos os números de 1 até esse número.

numero = int(input("digite um número:"))

contador = 1
soma = 0

while contador <= numero:
    soma = soma + contador
    contador = contador + 1
print(f"a soma de todos os números de 1 até {numero} é: {soma}")
```

```python
#Exercício 5 — Tabuada
#Peça um número ao usuário e mostre a tabuada desse número de 1 até 10 usando while.

n = int(input("digite um número: "))
contador = 1

while contador <= 10:
  resultado = n * contador
  print(f"{n} x {contador} = {resultado}")
  contador += 1
```

```python
#Exercício 6 — Senha
#Crie um programa que peça uma senha ao usuário.Enquanto a senha estiver incorreta, mostre:
#Senha incorreta! Quando o usuário digitar a senha correta, mostre:
#Acesso permitido! Considere que a senha correta seja "1234".

senha = 1234

a = int(input("digite a senha: "))

while a != senha:
  print(f"senha incorreta")
  a = int(input("digite a senha novamente: "))
print(f"acesso permitido")
```

```python
#Exercício 7 — Somando até digitar zero
#Faça um programa que peça vários números ao usuário. O programa deve continuar pedindo números até que o usuário digite 0.
#Ao final, mostre a soma de todos os números digitados.

soma = 0
numero = int(input("Digite um número (0 para parar): "))

while numero != 0:
    soma += numero
    numero = int(input("Digite um número (0 para parar): "))

print(f"A soma de todos os números é: {soma}")
```

```python
#Exercício 8 — Maior número
#Peça números ao usuário até que ele digite 0. Ao final, mostre qual foi o maior número digitado.

maior = 0

while True:
    numero = int(input("Digite um número (0 para sair): "))

    if numero == 0:
        break

    if numero > maior:
        maior = numero

print(f"O maior número digitado foi: {maior}")
```

```python
#Exercício 9 — Média das notas
#Faça um programa que peça 5 notas de um aluno usando while.
#Ao final, calcule e mostre a média das notas.

soma = 0
contador = 1

while contador <= 5:
    nota = float(input(f"Digite a {contador}ª nota: "))
    soma += nota
    contador += 1

media = soma / 5
print(f"A média das notas é: {media}")
```

```python
# Exercício 10 — Caixa eletrônico
# Faça um programa que comece com um saldo de R$ 1000,00.
# Mostre um menu:
# 1 - Depositar
# 2 - Sacar
# 3 - Ver saldo
# 4 - Sair

saldo = 1000.00

while True:
    print("\n--- Caixa Eletrônico ---")
    print("1 - Depositar")
    print("2 - Sacar")
    print("3 - Ver saldo")
    print("4 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        valor = float(input("Digite o valor para depositar: R$ "))
        if valor > 0:
            saldo += valor
            print(f"Depósito de R$ {valor:.2f} realizado com sucesso.")
        else:
            print("Valor inválido.")

    elif opcao == "2":
        valor = float(input("Digite o valor para sacar: R$ "))
        if 0 < valor <= saldo:
            saldo -= valor
            print(f"Saque de R$ {valor:.2f} realizado com sucesso.")
        else:
            print("Saldo insuficiente ou valor inválido.")

    elif opcao == "3":
        print(f"Seu saldo atual é: R$ {saldo:.2f}")

    elif opcao == "4":
        print("Saindo do sistema. Obrigado!")
        break

    else:
        print("Opção inválida. Escolha de 1 a 4.")
```
