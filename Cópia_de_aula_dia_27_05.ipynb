```python
# Jogo Menu Inteligente

# Criar um sistema que funcione até o usúario sair

# Menu

# 1 - Par/ Impar
# 2 - Positivo / Negativo
# 3 - Maior / Menor
# 4 - Sair

opcao = 0

while opcao != 4:
  print("")
  print("--- MENU ---")
  print("1 - Par / Impar")
  print("2 - Positivo / Negativo")
  print("3 - Maior")
  print("4 - Sair")

  opcao = int(input("escolha uma opção: "))

  if opcao == 1:
    num = int(input("digite um número: "))
    if num % 2 == 0:
      print(f"O número {num} é par")
    else:
      print(f"{num} é impar")

  elif opcao == 2:
    num = int(input("digite um número: "))

    if num > 0:
      print(f"o número {num} é positivo")

    else:
      print(f"o número é negativo")

  elif opcao == 3:
    a = int(input("digite um número: "))
    b = int(input("digite um número: "))

    if a > b:
      print(f"{a} é maior que {b}")

    elif a < b:
      print(f"{a} é menor que {b}")

    else:
      print("os números são iguais")

  elif opcao == 4:
    print("prorama encerrado!")
```

```python
import collections # Re-adding this import for tie-breaking logic

# Urna Eletrônica

# Criar um programa de votação com 3 candidatos

# Votar candidato A
# Votar candidato B
# Votar candidato C
# Votar Branco
# Votar Nulo
# Encerrar votação

# O sistema deve repetir até encerrar a votação
# Contar votos de cada candidato
# Contar votos brancos e nulos
# Ao encerrar, mostrar:

# Total de votos
# Total de votos por candidato
# Total de votos brancos
# Total de votos nulos
# Vencedor da eleição

# Regras extras

# Se houver empate entre candidatos: mostrar eleição empatada
# Se não ouver votos válidos, mostrar: nenhum candidato recebeu votos

votos_a = 0
votos_b = 0
votos_c = 0
votos_brancos = 0
votos_nulos = 0
# opcao = 0 # Not strictly needed with while True and break

while True:
  print("")
  print("")
  print("--- Urna Eletrônica ---")
  print("opção 1 - Votar no Candidato A")
  print("opção 2 - Votar no Candidato B")
  print("opção 3 - Votar no Candidato C")
  print("opção 4 - Voto Branco")
  print("opção 5 - Anular voto")
  print("opção 6 - Encerrar Votação")

  opcao =int(input(f"Escolha uma opção: "))

  if opcao == 1:
        votos_a += 1
        print(f"Você votou no Candidato A")

  elif opcao == 2:
        votos_b += 1
        print(f"Você votou no Candidato B")

  elif opcao == 3:
        votos_c += 1
        print(f"Você votou no Candidato C")

  elif opcao == 4:
        votos_brancos += 1
        print(f"Você votou Branco")

  elif opcao == 5:
        votos_nulos += 1
        print(f"Você anulou seu voto")

  elif opcao == 6:
        print(f"Votação encerrada!")

        break

# Cálculos finais
total_votos = votos_a + votos_b + votos_c + votos_brancos + votos_nulos
candidate_votes_list = [votos_a, votos_b, votos_c]

print("")
print("--- Resultado Final ---")
print(f"Total de Votos: {total_votos}")
print(f"Candidato A: {votos_a}")
print(f"Candidato B: {votos_b}")
print(f"Candidato C: {votos_c}")
print(f"Votos Brancos: {votos_brancos}")
print(f"Votos Nulos: {votos_nulos}")

# Regras Extras
if sum(candidate_votes_list) == 0:
    print("Nenhum candidato recebeu votos válidos.")
else:
    # Use collections.Counter for robust tie checking
    max_candidate_votes = max(candidate_votes_list)
    tied_candidates_count = collections.Counter(candidate_votes_list)[max_candidate_votes]

    if tied_candidates_count > 1:
        print("Eleição empatada entre os candidatos com mais votos.")
    else:
        if votos_a == max_candidate_votes:
            print("Vencedor: Candidato A")
        elif votos_b == max_candidate_votes:
            print("Vencedor: Candidato B")
        else:
            print("Vencedor: Candidato C")
```
