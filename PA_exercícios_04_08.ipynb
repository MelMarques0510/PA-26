```python
# Exercício 4 - Expressões aritméticas
# Calcule e mostre o resultado de:
# O dobro de 15
# A metade de 90
# O resto da divisão de 17 por 5
# 2 elevados à potência 4
# A média de 10, 8 e 6

a = 15 * 2
print(f"o dobro de 15 é igual a {a}")
("")

b = 90 / 2
print(f"a metade de 90 é igual a {b}")
("")

c = 17 % 5
print(f"o resto da divisão de 17 por 5 é igual a {c}")
("")

d = 2 ** 4
print(f"2 elevado à potência de 4 é igual a {d}")
("")

e = (10 + 8 + 6) / 3
print(f"a média de 10, 8 e 6 é igual a {e}")
```

```python
# Exercício 5 – Comparações (relacionais)
# Mostre o resultado (True ou False) das comparações:
# idade = 16
# nota = 8.5
# preco = 45.0

# idade >= 18
# nota > 7
# preco <= 50
# idade == 16
# nota != 10

idade = 16
nota = 8.5
preco = 45.0

print(f"idade = 18 {idade >= 18}")
print(f"nota = 8.5 {nota > 7}")
print(f"preco = 45.0 {preco <= 50}")
print(f"idade = 18 {idade == 16}")
print(f"nota = 8.5 {nota != 10}")
```
