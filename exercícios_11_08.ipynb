```python
# Exercício 1 – Idade para dirigir
#Receba a idade de uma pessoa e diga se ela pode tirar carteira de motorista.
#A pessoa pode tirar a carteira se tiver idade maior ou igual a 18 anos.

idade = int(input("digite sua idade: "))

if idade >= 18:
  print("você já pode tirar carteira de motorista!")

else:
  print("você ainda não pode tirar carteira de motorista!")
```

```python
# Exercício 2 – Par ou Ímpar
#Receba um número e diga se ele é par ou ímpar.

num = int(input("digite um número: "))

if num % 2 == 0:
  print(f"o número {num} é par")

else:
  print(f"o número {num} é impar")
```

```python
# Exercício 3 – Maior de dois números
#Receba dois números e mostre qual é o maior.
#Se os dois números forem iguais, diga que eles são iguais

num1 = int(input("digite o primeiro número: "))
num2 = int(input("digite o segundo número: "))

if num1 > num2:
  print(f"{num1} é maior que {num2}")

elif num1 < num2:
  print(f"{num1} é menor que {num2}")

else:
  print("os números são iguais")
```

```python
# Exercício 4 – Situação do aluno
#Receba a nota de um aluno e mostre sua situação:
# Nota maior ou igual a 7 → Aprovado
# Nota maior ou igual a 5 e menor que 7 → Recuperação
# Nota menor que 5 → Reprovado

nota = int(input("digite sua nota: "))

if nota >= 7:
  print("você foi aprovado!")

elif nota >=5:
  print("você está de recuperação!")

else:
  print("você está reprovado!")
```

```python
# Exercício 5 – Desconto na compra
#Receba o valor de uma compra.
#Se o valor for maior ou igual a 100, aplique 10% de desconto e mostre o valor final.
#Caso contrário, mostre o valor normal da compra.

v = float(input("digite o valor da sua compra: "))
desconto = v * 0.10
vf = v - desconto

if v >= 100:
  print(f"o valor final a ser pago é {vf}")

else:
  print(f"o valor final a ser pago é {v}")



```

```python
# Exercício 6 – Positivo, Negativo ou Zero
#Receba um número e diga se ele é positivo, negativo ou zero.

n = int(input("digite um número: "))

if n > 0:
  print(f"o número {n} é positivo")

elif n < 0:
  print(f"o número {n} é negativo")

else:
  print(f"o número é zero")
```

```python
# Exercício 7 – Classificação de idade
# Receba a idade de uma pessoa e classifique-a de acordo com sua faixa etária:
# Menor de 12 anos → Criança
# De 12 a 17 anos → Adolescente
# De 18 a 59 anos → Adulto
# 60 anos ou mais → Idoso

idade = int(input("digite sua idade: "))

if idade < 12:
  print("você é uma criança")

if idade >= 12 and idade <= 17:
  print("você é um adolescente")

elif idade >= 18 and idade <= 59:
  print("você é um adulto")

else:
  print("você é idoso")
```

```python
# Exercício 8 – Média de duas notas
#Receba duas notas, calcule a média e diga se o aluno foi aprovado ou reprovado.
#O aluno será aprovado se a média for maior ou igual a 7.

nota1 = float(input("digite sua primeira nota: "))
nota2 = float(input("digite sua segunda nota: "))
média = (nota1 + nota2) / 2

if média >= 7:
  print(f"sua média foi {média}, você foi aprovado!!")

else:
  print(f"sua média foi {média}, você foi reprovado!!")
```

```python
#Exercício 10 – Calculadora de IMC
#Receba o peso e a altura de uma pessoa, calcule o IMC usando a fórmula:
#IMC = peso / (altura × altura)
#Depois, classifique o resultado:
#IMC menor que 18.5 → Abaixo do peso
#IMC de 18.5 a 24.9 → Peso normal
#IMC de 25 a 29.9 → Sobrepeso
#IMC maior ou igual a 30 → Obesidade

peso = float(input("digite seu peso: "))
altura = float(input("digite sua altura: "))
IMC = peso / (altura * altura)

if IMC < 18.5:
  print("você está abaixo do peso")

elif IMC >= 18.5 and IMC <= 24.9:
  print("você está no peso normal")

elif IMC >= 25 and IMC <= 29.9:
  print("você está com sobrepeso")

else:
  print("você é obeso")
```
