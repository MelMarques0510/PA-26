```python
# QUESTÃO 01 – Contagem Crescente
# Elabore um programa que exiba na tela os números de 1 até 20, um em cada linha.

n = 20

for i in range(n):
  print(i+1)
```

```python
# QUESTÃO 02 – Soma dos Números
# Solicite ao usuário um número inteiro positivo. Utilizando um laço de repetição, calcule e exiba a soma de
# todos os números de 1 até o número informado.

n = int(input("Digite um número inteiro positivo: "))

soma = 0

for i in range(1, n + 1):
  soma += i

print(f"A soma dos números de 1 até {n} é: {soma}")
```

```python
# QUESTÃO 03 – Tabuada
# Solicite ao usuário um número inteiro e exiba sua tabuada de 1 até 10.

n = int(input("digite um número: "))

for i in range(1, 11):
  print(f"{n} x {i} = {n * i}")
```

```python
# QUESTÃO 04 – Fatorial
# Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.

n = int(input("digite um número: "))

if n < 0:
  print("não existe fatorial para números negativos")

elif n == 0:
  print("o fatorial de 0 é 1")

else:
  fatorial = 1
  for i in range(1, n + 1):
    fatorial *= i
  print(f"o fatorial de {n} é {fatorial}")
```

```python
# QUESTÃO 05 – Caixa Eletrônico
# Um caixa eletrônico possui notas de:
#R$ 100
#R$ 50
#R$ 20
#R$ 10
#R$ 5
#R$ 2
#R$ 1
#Solicite ao usuário o valor do saque e informe a menor quantidade possível de notas para realizar o pagamento.
# Requisitos:
# - Utilizar laços de repetição.
# - Não utilizar listas.
# - Exibir apenas as notas utilizadas.

valor_saque = int(input("Digite o valor do saque: R$ "))
valor_restante = valor_saque

print(f"Para o saque de R$ {valor_saque}, as notas serão:")

dominacoes = (100, 50, 20, 10, 5, 2, 1)

for nota in dominacoes:
  if valor_restante >= nota:
    quantidade = valor_restante // nota
    if quantidade > 0:
      print(f"{quantidade} nota(s) de R$ {nota}")
    valor_restante %= nota

if valor_restante > 0:
  print("Não foi possível sacar o valor exato com as notas disponíveis.")
```
